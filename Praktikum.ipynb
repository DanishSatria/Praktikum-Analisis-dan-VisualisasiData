```python
import pandas as pd

csv_path = '/content/data_praktikum_analisis_data.csv'

try:
    df = pd.read_csv(csv_path)
    print("Data loaded successfully. Here are the first 5 rows:")
    display(df.head())
except FileNotFoundError:
    print(f"Error: The file '{csv_path}' was not found. Please ensure it is uploaded correctly.")
except Exception as e:
    print(f"An error occurred while loading the CSV: {e}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a scatter plot to visualize the correlation
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Ad_Budget', y='Total_Sales', data=df)
plt.title('Correlation between Ad Budget and Total Sales')
plt.xlabel('Ad Budget')
plt.ylabel('Total Sales')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
